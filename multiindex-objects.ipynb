# Pivot Table
The pivot table takes simple column-wise data as input, and groups the entries into a two-dimenssional table that provides a multidimenssional summarization of the data.

```python
import numpy as np
import pandas as pd
import seaborn as sns
```

```python
df = sns.load_dataset('tips')
df.head()
```

```python
df.groupby('sex')[['total_bill']].mean()
```

```python
df.groupby(['sex','smoker'])[['total_bill']].mean().unstack()
```

```python
df.pivot_table(index='sex',columns='smoker',values='total_bill',aggfunc='std')
```
