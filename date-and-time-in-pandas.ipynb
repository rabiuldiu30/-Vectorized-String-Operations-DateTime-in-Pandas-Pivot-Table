# TimeStamp Object
Time stamps refrence particular moments in time(e,g. Oct 24th,2022 at 7.00pm)

Creating Timestamp Objects

```python
# creating a timestamp
type(pd.Timestamp('2023/1/5'))
```

```python
# variations
pd.Timestamp('2023-1-5')
pd.Timestamp('2023, 1, 5')
```

```python
# Only year
pd.Timestamp('2023')
```

```python
# using datetime.datetime object
import datetime as dt

x = pd.Timestamp(dt.datetime(2026,1,5,15,56,20))
x
```

```python
# fetching attributes
x.year
x.month
x.day
x.hour
x.minute
x.second
```
