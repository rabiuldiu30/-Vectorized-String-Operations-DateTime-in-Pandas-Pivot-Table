# TimeStamp Object
Time stamps refrence particular moments in time(e,g. Oct 24th,2022 at 7.00pm)

Creating Timestamp Objects

```python
# creating a timestamp
type(pd.Timestamp('2023/1/5'))
```

```python
# variations
pd.Timestamp('2023-1-5')
pd.Timestamp('2023, 1, 5')
```

```python
# Only year
pd.Timestamp('2023')
```

```python
# using datetime.datetime object
import datetime as dt

x = pd.Timestamp(dt.datetime(2026,1,5,15,56,20))
x
```

```python
# fetching attributes
x.year
x.month
x.day
x.hour
x.minute
x.second
```

```python
import numpy as np
date = np.array('2015-07-04', dtype=np.datetime64)
date
```

```python
date + np.arange(12)
```

# DatetimeIndex Object
A collection of pandas timestamp

```python
# from strings
type(pd.DatetimeIndex(['2023/1/1','2023/1/1','2021/1/1'])[0])
```

```python
# using python datetime object
pd.DatetimeIndex([dt.datetime(2023,1,1),dt.datetime(2022,1,1),dt.datetime(2021,1,1)])
```

```python
# using pd.timestamps
dt_index = pd.DatetimeIndex([pd.Timestamp(2023,1,1),pd.Timestamp(2022,1,1,),pd.Timestamp(2021,1,1)])
```

```python
# using datetimeindex as series index

pd.Series([1,2,3],index=dt_index)
```

```python
# generate daily dates in a given range
pd.date_range(start='2026/1/5', end='2026/2/28', freq='2D')
```

```python
pd.date_range(start='2026/1/5', end='2026/2/28', freq='3D')
```

```python
# B -> Business days
pd.date_range(start='2026/1/5', end='2026/2/28', freq='B')
```

```python
# W -> one week per day
pd.date_range(start='2026/1/5', end='2026/2/28', freq='W-FRI')
```

```python
# H -> Hourly data(factor)
pd.date_range(start='2026/1/5', end='2026/2/28', freq='6H')
```

```python
# M -> Month end
pd.date_range(start='2026/1/5', end='2026/2/28', freq='M')
```

```python
# MS -> Month start
pd.date_range(start='2026/1/5', end='2026/2/28', freq='MS')
```

```python
# A -> Year end
pd.date_range(start='2026/1/5', end='2026/2/28', freq='A')
```

```python
# using periods(number of result)
pd.date_range(start='2026/1/5', periods=25, freq='M')
```
